- mysql과 python을 연동한다
    1. 연동 기능이 존재하는 라이브러리(pymysql) 설치
    2. 라이브러리 로드
    3. 데이터베이스 서버와의 연결
        - 서버의 정보
            - 주소
            - port
            - 사용자 아이디
            - 비밀번호
            - 사용할 데이터베이스명(스키마)
    4. 커서(가상공간)를 생성
    5. query 문을 작성하여 질의를 한다.
        - insert, update, delete : return 존재 x
        - select : return이 존재
            - fetchall() : cursor에서 데이터를 가져온다(cursor에는 데이터가 삭제)
    6. DB server 와 Cursor 를 동기화
    7. DB server 와의 연결을 종료

```python
# 라이브러리 로드
import pymysql
```

```python
# DB 서버와의 연결 -> 열결되는 정보를 변수에 저장

_db = pymysql.connect(
    # host= 'localhost'
    host = '127.0.0.1', # 서버의 주소
    port = 3306,        # port 번호(무조건 숫자)
    user = 'root',      # 유저의 아이디
    password= '1234',   # 비밀번호(비밀번호가 없을시 '')
    db = 'multicam',

)
```

```python
# cryptography 에러 발생 시
# !pip install cryptography
```

```python
# _db를 이용하여 cursor 생성
# 기본값으로 커서 생성 -> select문을 이용하여 데이터를 로드할때 데이터의 타입이 tuple
cursor = _db.cursor()
# DictCursor 옵션 사용해서 커서 생성 -> select문을 이용하여 데이터를 로드할때 데이터의 타입이 dict
cursor2 = _db.cursor(pymysql.cursors.DictCursor)
```

```python
# query문을 작성해서 커서에게 질의 보낸다.
select_query = """
    select
    *
    from
    `emp`    
"""
# select_query를 커서에게 보내준다(cursor.execute())
cursor.execute(select_query)

cursor2.execute(select_query)
```

```python
cursor.fetchall()
```

```python
cursor2.fetchall()
```

```python
insert_query = """
    insert into
    `user_info`
    values
    ("test2", "1234", "park", "000101")
"""
# 커서에게 inser_query 질의를 보낸다
cursor2.execute(insert_query)
```

```python
select_query2 = """
    select*from `user_info`
""" # 미친 user를 uesr로 쓰고 앉아있네 ㅋ
cursor2.execute(select_query2)
cursor2.fetchall()
```

```python
# cursor와 DBserver와의 동기화
_db.commit()
```

```python
# 데이터베이스와의 연결을 종료
_db.close()
```

### 시계열 데이터
- 여러개의 데이터가 하나로 모여서 시계열 데이터 완성
- type(format)
    - 년도 표기(4자리수) : %Y
    - 년도 표기(2자리수) : %y
    - 월을 숫자로 표기 : %m
    - 월을 이름으로 표기 : %B
    - 월의 이름을 축약해서 표기 : %b
    - 일을 표기 : %d
    - 시를 24시 기준으로 표기 : %H
    - 시를 12시 기준으로 표기 : %I
    - AM/PM 표기 : %p
    - 분을 표기 : %M
    - 초를 표기 : %S
    - 요일을 이름으로 표기 : %A
    - 요일을 숫자로 표기 : %w ( 0 : 일 , 1 : 월, .... , 6 : 토)
    

```python
from datetime import datetime
```

```python
time_str = '2025-09-04'
```

```python
type(time_str)
```

```python
# 문자열 데이터를 시계열 데이터로 변환
# datetime 안에 있는 함수를 호출 -> 문자 데이터 시계열
# striptime()
    # 필수 매개변수 2개
        # 문자열 데이터 (시계열 데이터로 변경할 문자)
        # format -> 문자열 데이터가 어떠한 방식으로 이루어져 있는가?
time_data = datetime.strptime(
    time_str,
    "%Y-%m-%d"
)
```

```python
type(time_data)
```

```python
time_data
```

```python
datetime.strptime(
    '2025','%Y'
)
```

```python
# 시간 데이터를 문자로 변경
# strftype()
time_data.strftime(
    # "%d/%m/%Y,%A"
    "%w"
)
```

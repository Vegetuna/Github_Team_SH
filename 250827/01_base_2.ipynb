### 자료형 데이터
- 여러개의 데이터를 하나의 변수에 저장시키기 위한 형태

- tuple
    - 자료형 데이터 중에 가장 기본적인 형태
    - ()를 이용해서 데이터를 생성
    - 한번 생성이 되면 수정이 불가능하다(새로운 tuple을 생성)

- list
    - []를 이용해서 데이터 생성
    - tuple과 달리, 각 원소들을 수정하거나 삭제, 추가 작업이 가능 (리스트의 내부 함수 이용)
    - 원소의 개수를 0개부터 생성이 가능 (비어있는 리스트가 생성 -> 여러개의 데이터를 하나의 변수에 담을 준비 과정)

- dict
    - {}를 이요해서 데이터 생성
    - key : value 의 형태로 데이터 구성 (list에서는 위치를 기준으로 데이터를 추출한다면, dict는 key를 기준으로 데이터를 추출)
    - 원소의 개수는 0부터 생성이 가능

```python
# sort() -> 리스트 각각의 원소들을 비교하여 가장 작은 
# 데이터를 처음으로 가장 큰 데이터를 마지막으로 하여 정렬을 변경하는 함수

sample_list = [30, 10, 50, 7]
# sample_list에서 가장 작은 데이터를 담아줄 수 있는 새로운 빈 리스트 필요
sort_list = []
# sample_list의 길이만큼 반복을 하는 반복문을 생성
for i in range(len(sample_list)) :
    # 첫번째 원소의 데이터를 최솟값이라고 지정
    min_value = sample_list[0]
    
    # 두번째 원소(1)부터 마지막 원소(len)까지 min_value랑 비교해서 작은 데이터를 min_value에 대입
    for j in range(1, len(sample_list) ) :
        # min_value와 sample_list[j] 데이터를 비교하여 작은 값을 찾는다.
        if min_value > sample_list[j]:
            # min_value 가 리스트의 원소보다 크다면 실행
            min_value = sample_list[j]
            
    # print(min_value)
    sort_list.append(min_value)
    # sample_list 에서 최솟값으로 지정된 데이터를 sample_list에서 제거한다
    sample_list.remove(min_value)
```

```python
sort_list
```

```python
sample_list = [30, 10, 50, 7]
sort_list = []

for i in range(len(sample_list)):
    # 반복할때마다 sample_list 의 가장 작은 데이터를 찾는다 -> min()
    min_value = min(sample_list)
    print(min_value)
    sample_list.remove(min_value)
    sort_list.append(min_value)
print(sort_list)
```

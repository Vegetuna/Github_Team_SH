#### BART
- transform 기반의 모델
- Encoder, Decoder 모두 사용하는 모델
- 장문의 테스트에서 요약 데이터를 생성하는데 사용하는 모델
- Encoder : 입력이 되는 데이터를 BERT 형식으로 문장을 이해 (순방향, 역방향)
- Decoder : 출력이 되는 요약문의 GPT 형식으로 문장을 생성
- 해당 모델에서는 Tokenizer는 Sentencepeice 를 이용
- 인풋 tokenizer와 아웃풋 tokenizer는 따로 사용

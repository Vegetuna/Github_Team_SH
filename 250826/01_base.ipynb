```python
# while문
# 초기값 지정(시작값을 지정)
i = 1
while i <= 10:
    # i 가 10보다 작거나 같은 경우에는 반복 실행
    print(i)
    # i의 값이 거짓이 될 수 있도록 코드 작성 (조건식이 10보다 큰 경우가 거짓)
    # i의 값을 증가
    i += 1

print(i)
```

```python
# While문을 이용하여 1부터 10까지 합계 구하기
i = 1
result = 0

while i <= 10:
    # result에 i를 더하고 result에 결과를 대입
    result += i
    # i에 1을 더하고 i에 대입
    # 첫번째 반복 : i(1) -> i = 1 + 1 -> (2)
    i += 1
    # 반복 실행되는 코드가 없으므로 while문으로 되돌아감

print(result)
```

```python
# While문을 이용하여 1부터 10까지 합계 구하기
i = 1
result = 0

while i <= 10:
    # 첫번째 반복 : i(1) -> i + 1 -> i(2)
    i += 1
    # 첫번째 반복 : i(2), result(0) -> result = 0+2 -> result(2)
    result += i
    
print(result)
```

```python
# 반복문 안에서 조건식을 이용하여 break를 이용
# 1부터 시작해서 1씩 증가시킨 데이터를 누적합했을때
# 누적합의 값이 7000이 넘어가는 시점은 언제인가
# case1 (for문)
# 누적합을 해줄 공간을 형성
result = 0

for i in range(1,1000,1):
    # i를 result에 누적합
    # 118번째 반복
        # 1 조건식이 존재하는 경우 : result가 아직 7000 미만 -> break 실행 x
        # i(118), result(6903) -> result = 6903 + 118 -> result(7021)
        # 2 조건식 존재하는 경우 : result 가 7000 이상 -> break 실행 o
    # 2 조건식이 존재하는 경우 : result가 7000 이상 -> break 실행 o
    # 119번째 반복
        # 1 조건식 존재하는 경우 : result 가 7000 이상 -> break 실행 o -> result(7021)
        # 2 조건식 존재하는 경우 : 해당반복 실행 x

    # 1 조건식
    # if result > 7000:
        # break
    result += i
    # 2 조건식
    if result > 7000:
        break
    
print(result)
print(i)
```

```python
# while문을 이용하여 1부터 1씩 증가시킨 데이터의 누적합이 8400이 넘어가는
# 시점을 구하라

# 시작값 설정
result = 0
# 누적합 초기값 지정
i = 1
# while 조건식(누적합이 8400이상인 경우 반복문을 종료 추가)
# 반복 횟수가 명확하지 않을때 -> 무한반복
while True:
    # 시작값 누적합에 더하고 대입
    result += i
    if result >= 8400:
        break
    # 시작값을 증가
    i += 1
# 시작값과 누적값의 값을 출력
print(result)
print(i)
```

```python
# continue 문 : 반복문으로 되돌아간다. (반복 실행할 코드가 끝나면 같은 행동)
for i in range(10):
    continue # continue가 실행이 되면, 밑의 반복 실행되는 코드를 무시하고
             # 반복문으로 돌아간다. 반복 횟수는 증가.
    print(i)

print(i)
```

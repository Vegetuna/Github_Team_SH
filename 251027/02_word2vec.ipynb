```python
from gensim.models import Word2Vec
```

```python
# 토큰화가 된 데이터셋을 임의로 생성
sentences = [
    ['오늘', '날씨', '좋다'],
    ['강아지','산책','기분','좋다'],
    ['비','오다','산책','못하다']
]
```

```python
# Word2Vec 객체를 생성
model = Word2Vec(
    sentences= sentences,
    vector_size= 100,
    window= 5,
    min_count=1,
    sg= 1,
    epochs=100,
    seed= 42
)
```

```python
# 단위 벡터 출력
print(model.wv['산책'])
```

```python
# 코사인 유사도
print(model.wv.most_similar('산책'))
```

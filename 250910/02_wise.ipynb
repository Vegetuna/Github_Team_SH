```python
import pandas as pd
from bs4 import BeautifulSoup as bs
from selenium import webdriver
import requests
```

```python
# 접속 주소
url = 'https://comp.wisereport.co.kr/company/c1010001.aspx?cmp_cd='
# 종목 코드
input_code = '000660'
```

### 연습
- webdriver의 Chrome을 이용하여 url + input_code 로 요청
- driver의 html 문서를 불러온다
- 불러온 html을 BeautifulSoup을 이용해서 데이터 파싱
- div 태그 중 class의 이름이 "cmp_comment"와 같은 태그를 선택

```python
input_code = '005180'
```

```python
res = requests.get(url+input_code)
```

```python
res
```

```python
soup = bs(res.text, 'html.parser')
```

```python
cmp_info = list(
    map(
        lambda x : x.get_text(),
        soup.find(
            'div', attrs= {'class' : 'cmp_comment'}
        ).find_all('li')
    )
)
cmp_etc = list(
    map(
        lambda x : x.get_text(),
        soup.find(
            'div', attrs= {'class' : 'cmp_comment_etc'}
        ).find_all('li')
    )
)
```

```python
df3 = pd.DataFrame(
    {
        'cmp_info' : cmp_info,
        'cmp_etc' : cmp_etc
    }
)
df3['code'] = input_code
```

```python
df3
```

```python
pd.concat(
    [df2,df3], axis = 0
)
```

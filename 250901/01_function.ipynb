    - n 팩터리얼의 값을 되돌려주는 함수

```python
# 함수 생성 -> 매개변수, 함수의 결과에 대한 힌트를 포함
def func_2( n : int ) -> int :
    # 1부터 시작해서 자기 자신까지 1씩 증가시키면서 누적 곱
    result = 1
    for i in range(2, n+1 , 1):
        result *= i
    return result
```

```python
func_2(5)
```

```python
def func_3( n : int)->int :
    result = n
    for i in range(n-1,1,-1):
        result*=i
    return result
```

```python
# list에 내장된 reverse() 함수를 생성
def list_reverse(_x):
    # 가장 마지막의 데이터를 추출하여 새로운 리스트의 추가
    result = []
    
    for i in range(1,len(_x) , -1):
        result.append(
            _x[-i]
        )
    return result
```

```python
# find(), index() 함수를 모른다는 기준으로 함수 구성
def c_find(_data, _word):
    # index들이 저장될 공간을 설정
    result = []
    result2 = ()
    for i in range (len(_data)) :
        # _data가 문자인가? -> 변수의 타입이 무엇인가 type()
        if type(_data) == str :
            flag = _data[i:i+len(_word)] == _word
        else :
            flag = _data[i] == _word

        if flag:
            # list 데이터에서 데이터를 추가할때
            result.append(i)
            # result2는 tuple에 데이터를 추가?
            result2 += (i,) # 덮어쓰기 느낌?
    return result, result2
```

```python
def sosu(_start=2, _end=100):
    # 소수를 담을 리스트와 반복 횟수를 셀 변수를 초기화합니다.
    result = []
    total_iterations = 0
    
    # 1. 시작 값을 2 미만으로 설정하면 2로 보정합니다.
    if _start < 2:
        print("2이상의 정수만 입력해.")
    
    # _start 부터 _end까지의 수를 반복합니다.
    number = _start
    while number <= _end:
        is_prime = True
        j = 2
        
        # 2. 2부터 number-1 까지 반복하며 나누어 떨어지는지 확인합니다.
        # 소수는 1과 자신 외에 나누어 떨어지는 수가 없습니다.
        while j < number:
            total_iterations += 1  # 나눗셈 연산 횟수를 셉니다.
            if number % j == 0:
                is_prime = False   # 나누어 떨어지면 소수가 아님
                break              # 더 이상 검사할 필요 없으므로 내부 반복문 탈출
            j += 1
            
        # 3. is_prime이 True인 경우, 즉 소수인 경우에만 리스트에 추가합니다.
        if is_prime:
            result.append(number)
            
        # 다음 숫자로 넘어갑니다.
        number += 1
        
    return result, total_iterations
```

```python
def get_primes_optimized(_start=2, _end=100):
    """
    _start부터 _end까지의 범위에서 가장 효율적으로 소수를 찾아 반환합니다.

    Parameters:
        _start (int): 소수를 찾을 시작 숫자 (기본값 2)
        _end (int): 소수를 찾을 마지막 숫자 (기본값 100)
    
    Returns:
        tuple: (소수 리스트, 총 반복 횟수)
    """
    prime_numbers = []
    total_iterations = 0

    if _start < 2:
        _start = 2

    for number in range(_start, _end + 1):
        is_prime = True
        
        # 최적화 1: 2로 나누어 떨어지는지 먼저 확인 (짝수 제외)
        if number % 2 == 0 and number > 2:
            is_prime = False
        else:
            # 최적화 2: 제곱근까지만 확인
            # 어떤 수의 약수는 그 수의 제곱근을 넘지 않기 때문에,
            # 제곱근까지만 확인하면 충분합니다.
            limit = int(number**0.5) + 1
            
            # 홀수만 검사 (2는 이미 처리했기 때문)
            for i in range(3, limit, 2):
                total_iterations += 1
                if number % i == 0:
                    is_prime = False
                    break

        if is_prime:
            prime_numbers.append(number)
    
    return prime_numbers, total_iterations
```

```python
get_primes_optimized(2,100)
```

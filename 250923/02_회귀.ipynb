# 회귀분석
- 회귀 분석은 실제값과 예측값의 차이를 기반으로 한 지표를 중심으로 성능 평가 지표
- 실제값과 예측값의 차이를 구해서 이것들을 단순 합산하면 양수와 음수가 섞여있기 때문에 문제 발생

### MAE
- 실제값과 예측값의 차이를 제곱해 평균을 구한 값
- 에러의 크기고 그대로임 반영
- 이상치에 영향을 준다

### MSE
- 실제값과 예측값의 차이의 제곱을 한 값의 의미
- 실제갑소가 예측값의 사이들 제곱을 한 값의 평균을 구한값
- 실제값과 예측값의 면적합을 의미
- 이상치에 영향을 받음

### RMSE
- MSE 값에 루트
- 에러에 제곱을 하면 에러가 클수록 그에 따른 가중치가 높이 반영
- 손실이 기하급수적으로 증가하는 상황에서 실제 오류 평균보다 값이 커지는 현상을 상쇄함

### MSLE
- 실제값과 예측값의 차이를 제곱해 평균을 한 값(MSE)에 로그를 적용한 값
- RMSE와 같이 손실이 기하급수적으로 증가하는 상황에서 실제 오류 평균보다 값이 커지는 현상을 상쇄하기 위해 사용

### MAPE
- MAE를 퍼센트로 표시
- 오차가 예측값에서 차지하는 정도를 나타냄

- 회귀 분석의 순서
1. 모델 선택하고 import
2. 모델 객체(class)를 생성
3. 모델을 학습
4. 데이터(독립 변수)를 주고 예측
5. 실제 값과 예측 값을 확인하고 오차를 체크

```python
# csv 폴더 안에 있는 boston.csv 파일 로드
boston = pd.read_csv('./../csv/boston.csv')
boston.head()
```

    - PTRATIO : 지역별 학생 - 교사 비율
    - B : 인종 지수 (흑인 거주 비율)
    - LSTAT : 저소득층 비율(%)
    - Price : 주택의 가격의 중앙값

```python
# subplots를 이용하여 Price와 CHAS 컬럼을 제외한 나머지 컬럼들 간의 산점도 그래프
fig, axes = plt.subplots(nrows=3, ncols=4, figsize=(16, 10))

for i,column in zip(range(len(cols)), cols):
    # row = i//4
    row = int(i/4)
    col = i % 4

    sns.regplot(x = col, y = boston['Price'], data= boston, ax= axes[row][col])

plt.show()
```

```python
# 모든 컬럼을 기준으로 train, test 데이터셋으로 데이터 분할
X = boston.drop('Price', axis=1).values
Y = boston['Price'].values
```

```python
from sklearn.model_selection import train_test_split
```

```python
boston['CHAS'].value_counts()
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state= 100
)
```

```python
# 종속변수의 학습 데이터와 평가 데이터의 평균을 확인
print(Y_train.mean())
print(Y_test.mean())
```

```python
# 모델의 선택
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
# class(객체)를 생성
```

```python
# 모델의 학습 -> X_train, Y_train
lr.fit(X_train, Y_train)
```

```python
# 예측 -> 독립 변수중 평가데이터로 구성된 X_test 데이터를 대입
pred = lr.predict(X_test)
```

```python
pred_df = pd.DataFrame(pred, columns=['Pred price'])
actual_df = pd.DataFrame(Y_test, columns=['Actual prince'])
```

```python
test_df = pd.DataFrame(X_test, columns=boston.columns.difference(['Price']))
```

```python
# 두개의 데이터프레임을 결합
pd.concat(
    [test_df,pred_df, actual_df],
    axis=1
)
```

```python
# 평가 지표 생성
import numpy as np
from sklearn.metrics import mean_absolute_error, \
    mean_squared_error, mean_squared_log_error, \
    mean_absolute_percentage_error, root_mean_squared_error
```

```python
mae = mean_absolute_error(Y_test,pred)
mse = mean_squared_error(Y_test, pred)
msle = mean_squared_log_error(Y_test, pred)
mape = mean_absolute_percentage_error(Y_test, pred)
rmse = root_mean_squared_error(Y_test, pred)
rmse2 = np.sqrt(mse)
```

```python
print('MAE : ', round(mae, 2))
print('MSE : ', round(mse, 2))
print('RMSE : ', round(rmse, 2))
print('RMSE2 : ', round(rmse2, 2))
print('MSLE : ', round(msle, 2))
print('MAPE : ', round(mape, 2)*100)
```

```python
from sklearn.metrics import r2_score

# 결정 계수 출력
r2 = r2_score( Y_test, pred )

print("R2 Score : ", round(r2, 2))
```

```python
# R2 - score의 문제 : 컬럼(피쳐)의 개수가 많아지면 R2 - Score가 증가하는 경향
# 수정된 R2 - Score 생성
# 1 - ( (1 - r2)*(n - 1)/(n - p - 1) )

# X_test의 인데스의 수
n = X_test.shape[0]

# X_test의 컬럼의 수
p = X_test.shape[1]

adj_r2 = 1 - ( (1 - r2)* (n - 1) / ( n - p - 1))

print('Adj R2 Score :', round(adj_r2, 2))
```

```python
# 회귀계수 출력
print(lr.coef_)
```

```python
pd.Series(
    lr.coef_,
    index = boston.columns.difference(['Price'])
)
```

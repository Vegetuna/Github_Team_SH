### 문자열에서 제공하는 내장 함수
- upper() : 문자열을 모두 대문자로 변경
- lower() : 문자열을 모두 소문자로 변경
- swapcase() : 대문자를 소문자로, 소문자를 대문자로 변경
- title() : 단어의 첫글자만 대문자로 변경
- strip() : 기본적으로는 문자열의 시작과 끝의 존재하는 공백을 제거
    - rstrip() : 문자열의 오른쪽 끝 공백 제거
    - lstrip() : 문자열의 왼쪽 끝 공백 제거
- replace() : 특정 문자를 찾아서 다른 문자로 치환
- split() : 특정 문자를 찾아서 문자를 잘라주는 기능 -> 결과 list

```python
# 인자의 개수가 가변인 경우
def func_7( *args ):
    # args의 타입? -> tuple
    print(type(args))
    # 인자들의 누적 합을 구해서 되돌려주는 기능
    # 합계 초기값 지정
    result = 0
    # agrs의 원소의 개수만큼 반복 실행하는 반복문 생성
    for i in args :
        # i 는 args의 원소들이 각각 대입
        result += i
    
    # return은 값을 되돌려주고 함수를 종료
    
    return result

```

```python
func_7(1, 2, 3, 4, 5)
```

```python
# 인자의 개수가 가변인 매개변수와 개수가 1개인 매개변수가 존재하는 함수
def func_9(*args, data) :
    result = 0
    # data가 1이라면 누적합을 되돌려주고
    if data == 1:
        result = 0
        for i in args :
            result += i

    # data가 0이라면 누적곱을 되돌려준다.
    if data == 0 :
        result = 1
        for i in args:
            result *= i
    return result
        
```

```python
func_9(1,2,3,4,5, data = 1)
# 가변의 인자가 첫번째로 들어왔을때 error
```

### 연습
- 함수를 생성
    - 매개변수 2개
        - _datas : 인자의 개수가 가변인 매개변수
        - _type : 기본값이 설정 ('max')
    - _type이 'max' 라면 _datas 들 중에 가장 큰 값을 되돌려준다.
    - _type이 'min' 이라면 _datas 들 중에 가장 작은 값을 되돌려준다.
    - _type이 'meas' 이라면 _datas 들의 평균을 되돌려준다.
    - 위에 3가지 경우가 아니라면 "_type 이 맞지 않습니다" 를 되돌려준다.


```python
def func_practice(*_datas, _type = 'max'):
    if _type == 'max' :
        return max(_datas)
    if _type == 'min' :
        return min(_datas)
    if _type == 'mean':
        return sum(_datas)/len(_datas)
    return "_type이 맞지 않습니다"
    
```

```python
func_practice(1,2,3,4, _type = 1)
```

```python
def func_12(*_datas, _type = 'max') :
    if _type == 'max' :
        result = max(_datas)
    elif _type == 'min':
        result = min(_datas)
    elif _type == 'mean':
        result = sum(_datas)/len(_datas)
    else :
        result = '_type 이 맞지 않습니다'
    return result
```

```python
func_12(1,2,3,4,5)
```

```python
func_12(1,2,3,4,5, _type = 'mean')
```

### class 선언
    - class 키워드를 이용해서 class를 선언
    - class 이름은 일반적으로 단어의 첫글짜를 대문자로 표기
### class 생성
    - 선언이 된 class를 특정 공간에 저장하는 과정(변수를 생성해서 class 를 대입)
    - 생성된 class 안에는 변수와 함수가 존재(같은 변수명을 사용해도 각각 저장)
### class 내부의 변수, 함수를 호출
    - class가 저장되어있는 공간(변수명) + ~안에(.) + class에 존재하는 변수(변수명), 함수(함수명()) 호출

```python
# 가장 기본적인 class 선언
class Class_1 :
    # 생성자 함수(초기화 함수)
    # class 내부에 함수를 생성할때는 일반적으로 매개변수 self는 기본
    # self : 자기자신(생성이 되는 class 공간)
    # 따로 호출할 필요 없음
    def __init__(self, _a, _b):
        # class 내부에서 사용할 변수들을 저장하는 기능
        # self.a -> self(자기자신) + .(안에) + a(변수명)
        self.a = _a
        self.b = _b
        # __init__ 함수를 class 내부에서 사용할 데이터를 저장하는 기능
        # return 따로 만들지 않는다.

    def add_value(self):
        result = self.a + self.b
        return result
        
    
```

```python
# Class 생성
# Class_1 을 생성 -> __init__ 함수 호출
# -> __init__ 함수는 매개변수 3개 -> 입력값을 3개
test = Class_1(10, 7)
```

```python
# 3번째 인자는 저장될 수 있는 공간이 존재하지 않는다 -> error
# test2 = Class_1(10, 7, 1)
#  __init__(self, _a, _b)에서
# self는 자기 자신의 위치값을 가지고 있음 -> 이미 데이터가 있다.(자동 기입이 됨)
```

```python
test2 = Class_1(20, 4)
```

```python
# class 내부에 있는 변수를 호출
# 생성이 된 클래스의 위치(변수명) + 안에(.) + 변수명
print(test.a)
print(test2.a)
```

```python
print(test.add_value())
```

```python
# class 내부에 있는 함수를 호출
# 생성이 된 클래스의 위치(변수명) + 안에(.) + 함수()
# add_value() 함수를 호출
test.add_value()
```

```python
# 계산기 class 선언
class Calculate :

    # 생성자 함수 -> 두개의 숫자 데이터를 저장하기 위한 기능
    # 생성자 함수 필수 x
    def __init__(self, _a, _b):
        self.a = _a
        self.b = _b
    
    # 두개의 데이터를 더한 뒤 되돌려주는 함수
    def add(self):
        result = self.a + self.b
        return result
    # 두개의 데이터를 뺀 뒤 되돌려주는 함수
    def sub(self):
        result = self.a - self.b
        return result
    # 두개의 데이터를 곱한 뒤 되돌려주는 함수
    def multi(self) :
        result = self.a * self.b
        return result
    # 두개의 데이터를 나눈 뒤 되돌려주는 함수
    def dev(self):
        result = self.a / self.b
        return result
    
```

```python
cal_1 = Calculate(20,4)
cal_2 = Calculate(5,10)
```

```python
print(cal_1.add(), cal_1.sub(), cal_1.multi(),cal_1.dev())
```

```python
print(cal_2.add(), cal_2.sub(), cal_2.multi(),cal_2.dev())
```

```python
# cal_1, cal_2 함수들을 호출할 때 출력이 나온다는 의미는 
# class 내부의 데이터가 변경이 되지 않는다.
```

```python
# class 내부의 변수에 데이터를 변경하는 class 선언
class Class_2 :
    def __init__(self, _a):
        self.a = _a

    # self.a 연산(더하기)하며 변경하는 함수 -> 매개변수 2개 추가.
    def func_1(self, _val, inplace = False) :
        # inplace가 True 라면 -> self.a에 _val를 더하고 self.a를 변경
        if inplace :
            self.a += _val
        # inplace가 False 라면 -> self.a에 _val를 더한 데이터를 되돌려준다
        else:
            result = self.a + _val
            return result
```

```python
test3 = Class_2(10)
```

```python
# test3 내부에 있는 func_1()함수를 호출
# 해당 함수에는 매개변수 3개 -> self(test3의 위치가 자동 입력), 
#                           _val(필수입력), 
#                           inplace(기본값이 False)
test3.func_1(5)
```

```python
test3.func_1(5, inplace=True)
```

```python
test3.func_1(5)
```

```python
# class 변수
# 같은 class를 생성하는 경우 서로 공유하는 변수
class Class_3 :
    # self 변수들의 생성위치 -> 함수 안에서 생성
    # class 변수는 생성위치 함수 밖에서 생성
    cnt = 0
    users = []


    def add_user(self, _name):
        
        # 유저가 등록됨에 따라서 cnt는 1씩 증가
        # self.cnt += 1 하면 카운트 안됨, 지역변수 공유를 못해서 클래스 위치를 지정해줘야함
        Class_3.cnt += 1
        # 유저의 이름은 users에 추가
        self.users.append(_name)
        
```

```python
user1 = Class_3()
# Class_3() 빈칸 안에 값 없는 이유 -> 생성자 함수를 선언하지 않았기 때문에
user2 = Class_3()
```

```python
user1.add_user('kim')
user2.add_user('park')
```

```python
print(user1.cnt, user1.users)
print(user2.cnt, user2.users)
```
